```python
import torch
import torch.nn as nn
import torch.optim as optim
import torch.nn.functional as F
import torchvision
from torchvision import datasets, transforms
from torchvision.models import mobilenet_v2, MobileNet_V2_Weights
import matplotlib.pyplot as plt
import numpy as np

# --- Device Configuration ---
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")

# =====================================================================
# PART 1: Data Preparation & Visualization
# =====================================================================
vis_transform = transforms.ToTensor()
vis_dataset = datasets.STL10(root='./data', split='train', download=True, transform=vis_transform)
classes = ['airplane', 'bird', 'car', 'cat', 'deer', 'dog', 'horse', 'monkey', 'ship', 'truck']

def visualize_samples(dataset, classes):
    fig, axes = plt.subplots(10, 4, figsize=(10, 20))
    class_counts = {i: 0 for i in range(10)}
    for img, label in dataset:
        if class_counts[label] < 4:
            col = class_counts[label]
            img_np = img.permute(1, 2, 0).numpy()
            axes[label, col].imshow(img_np)
            axes[label, col].axis('off')
            if col == 0:
                axes[label, col].set_title(classes[label], loc='left')
            class_counts[label] += 1
        if all(count == 4 for count in class_counts.values()):
            break
    plt.tight_layout()
    plt.show()

# Show dataset samples
visualize_samples(vis_dataset, classes)

# --- Data Augmentations ---
train_transform = transforms.Compose([
    transforms.RandomCrop(64),
    transforms.RandomHorizontalFlip(),
    transforms.ColorJitter(brightness=0.2, contrast=0.2),
    transforms.ToTensor(),
    transforms.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5))
])

test_transform = transforms.Compose([
    transforms.CenterCrop(64),
    transforms.ToTensor(),
    transforms.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5))
])

full_train_dataset = datasets.STL10(root='./data', split='train', download=True, transform=train_transform)
test_dataset = datasets.STL10(root='./data', split='test', download=True, transform=test_transform)

train_size = int(0.8 * len(full_train_dataset))
val_size = len(full_train_dataset) - train_size
train_dataset, val_dataset = torch.utils.data.random_split(full_train_dataset, [train_size, val_size])

train_loader = torch.utils.data.DataLoader(train_dataset, batch_size=64, shuffle=True)
val_loader = torch.utils.data.DataLoader(val_dataset, batch_size=64, shuffle=False)
test_loader = torch.utils.data.DataLoader(test_dataset, batch_size=64, shuffle=False)

# =====================================================================
# PART 2: Model Architectures
# =====================================================================
# 1. Logistic Regression
class LogisticRegressionModel(nn.Module):
    def __init__(self):
        super(LogisticRegressionModel, self).__init__()
        self.linear = nn.Linear(3 * 64 * 64, 10)

    def forward(self, x):
        x = x.view(x.size(0), -1)
        return self.linear(x)

# 2. Fully Connected NN
class FullyConnectedNN(nn.Module):
    def __init__(self):
        super(FullyConnectedNN, self).__init__()
        input_size = 3 * 64 * 64
        self.flatten = nn.Flatten()
        self.fc1 = nn.Linear(input_size, 1024)
        self.bn1 = nn.BatchNorm1d(1024)
        self.drop1 = nn.Dropout(0.5)
        self.fc2 = nn.Linear(1024, 512)
        self.bn2 = nn.BatchNorm1d(512)
        self.drop2 = nn.Dropout(0.5)
        self.fc3 = nn.Linear(512, 256)
        self.bn3 = nn.BatchNorm1d(256)
        self.drop3 = nn.Dropout(0.5)
        self.classifier = nn.Linear(256, 10)

    def forward(self, x):
        x = self.flatten(x)
        x = self.drop1(F.relu(self.bn1(self.fc1(x))))
        x = self.drop2(F.relu(self.bn2(self.fc2(x))))
        x = self.drop3(F.relu(self.bn3(self.fc3(x))))
        return self.classifier(x)

# 3. Custom CNN
class CustomCNN(nn.Module):
    def __init__(self):
        super(CustomCNN, self).__init__()
        self.conv1 = nn.Conv2d(3, 32, kernel_size=3, padding=1)
        self.bn1 = nn.BatchNorm2d(32)
        self.pool1 = nn.MaxPool2d(kernel_size=2, stride=2)
        self.conv2 = nn.Conv2d(32, 64, kernel_size=3, padding=1)
        self.bn2 = nn.BatchNorm2d(64)
        self.pool2 = nn.MaxPool2d(kernel_size=2, stride=2)
        self.flatten = nn.Flatten()
        self.fc1 = nn.Linear(64 * 16 * 16, 512)
        self.drop1 = nn.Dropout(0.5)
        self.fc2 = nn.Linear(512, 128)
        self.drop2 = nn.Dropout(0.5)
        self.classifier = nn.Linear(128, 10)

    def forward(self, x):
        x = self.pool1(F.relu(self.bn1(self.conv1(x))))
        x = self.pool2(F.relu(self.bn2(self.conv2(x))))
        x = self.flatten(x)
        x = self.drop1(F.relu(self.fc1(x)))
        x = self.drop2(F.relu(self.fc2(x)))
        return self.classifier(x)

# 4 & 5. MobileNetV2
def get_mobilenet_model(freeze_features=True):
    model = mobilenet_v2(weights=MobileNet_V2_Weights.DEFAULT)
    for param in model.features.parameters():
        param.requires_grad = not freeze_features
    model.classifier = nn.Sequential(
        nn.Dropout(0.2),
        nn.Linear(1280, 512),
        nn.ReLU(),
        nn.Dropout(0.5),
        nn.Linear(512, 128),
        nn.ReLU(),
        nn.Linear(128, 10)
    )
    return model

# =====================================================================
# PART 3: Training & Evaluation Functions
# =====================================================================
def count_parameters(model):
    return sum(p.numel() for p in model.parameters() if p.requires_grad)

def train_and_eval_model(model, train_loader, val_loader, test_loader, epochs=10, lr=0.001, weight_decay=0):
    model = model.to(device)
    criterion = nn.CrossEntropyLoss()
    optimizer = optim.Adam(model.parameters(), lr=lr, weight_decay=weight_decay)
    history = {'train_loss': [], 'val_loss': [], 'train_acc': [], 'val_acc': []}

    for epoch in range(epochs):
        model.train()
        train_loss, correct_train, total_train = 0.0, 0, 0
        for inputs, labels in train_loader:
            inputs, labels = inputs.to(device), labels.to(device)
            optimizer.zero_grad()
            outputs = model(inputs)
            loss = criterion(outputs, labels)
            loss.backward()
            optimizer.step()
            train_loss += loss.item() * inputs.size(0)
            _, predicted = torch.max(outputs.data, 1)
            total_train += labels.size(0)
            correct_train += (predicted == labels).sum().item()

        epoch_train_loss = train_loss / len(train_loader.dataset)
        epoch_train_acc = 100 * correct_train / total_train

        model.eval()
        val_loss, correct_val, total_val = 0.0, 0, 0
        with torch.no_grad():
            for inputs, labels in val_loader:
                inputs, labels = inputs.to(device), labels.to(device)
                outputs = model(inputs)
                loss = criterion(outputs, labels)
                val_loss += loss.item() * inputs.size(0)
                _, predicted = torch.max(outputs.data, 1)
                total_val += labels.size(0)
                correct_val += (predicted == labels).sum().item()

        epoch_val_loss = val_loss / len(val_loader.dataset)
        epoch_val_acc = 100 * correct_val / total_val
        history['train_loss'].append(epoch_train_loss)
        history['val_loss'].append(epoch_val_loss)
        history['train_acc'].append(epoch_train_acc)
        history['val_acc'].append(epoch_val_acc)
        print(f"Epoch {epoch+1:02d}/{epochs} | Train Loss: {epoch_train_loss:.4f} Acc: {epoch_train_acc:.2f}% | Val Loss: {epoch_val_loss:.4f} Acc: {epoch_val_acc:.2f}%")

    model.eval()
    correct_test, total_test = 0, 0
    with torch.no_grad():
        for inputs, labels in test_loader:
            inputs, labels = inputs.to(device), labels.to(device)
            outputs = model(inputs)
            _, predicted = torch.max(outputs.data, 1)
            total_test += labels.size(0)
            correct_test += (predicted == labels).sum().item()
    test_acc = 100 * correct_test / total_test
    print(f"==== Test Accuracy: {test_acc:.2f}% ====\n")
    return history, test_acc

def plot_results(history, title):
    epochs = range(1, len(history['train_loss']) + 1)
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
    ax1.plot(epochs, history['train_loss'], label='Train Loss', marker='o')
    ax1.plot(epochs, history['val_loss'], label='Validation Loss', marker='o')
    ax1.set_title(f'{title} - Loss')
    ax1.set_xlabel('Epochs')
    ax1.set_ylabel('Loss')
    ax1.legend()
    ax1.grid(True)
    ax2.plot(epochs, history['train_acc'], label='Train Accuracy', marker='o')
    ax2.plot(epochs, history['val_acc'], label='Validation Accuracy', marker='o')
    ax2.set_title(f'{title} - Accuracy')
    ax2.set_xlabel('Epochs')
    ax2.set_ylabel('Accuracy (%)')
    ax2.legend()
    ax2.grid(True)
    plt.show()

# =====================================================================
# PART 4: Execute Experiments
# =====================================================================
print("\n--- Model 1: Logistic Regression ---")
model1 = LogisticRegressionModel()
hist1, test1 = train_and_eval_model(model1, train_loader, val_loader, test_loader, epochs=10, lr=0.001)
plot_results(hist1, "Model 1: Logistic Regression")

print("\n--- Model 2: Fully Connected NN ---")
model2 = FullyConnectedNN()
print(f"Learnable Parameters: {count_parameters(model2):,}")
hist2, test2 = train_and_eval_model(model2, train_loader, val_loader, test_loader, epochs=15, lr=0.001, weight_decay=1e-4)
plot_results(hist2, "Model 2: Fully Connected NN")

print("\n--- Model 3: CNN ---")
model3 = CustomCNN()
print(f"Learnable Parameters: {count_parameters(model3):,}")
hist3, test3 = train_and_eval_model(model3, train_loader, val_loader, test_loader, epochs=15, lr=0.001, weight_decay=1e-4)
plot_results(hist3, "Model 3: CNN")

print("\n--- Model 4: MobileNetV2 Fixed ---")
model4 = get_mobilenet_model(freeze_features=True)
hist4, test4 = train_and_eval_model(model4, train_loader, val_loader, test_loader, epochs=10, lr=0.001)
plot_results(hist4, "Model 4: MobileNetV2 Fixed")

print("\n--- Model 5: MobileNetV2 Learned (Fine-Tuning) ---")
model5 = get_mobilenet_model(freeze_features=False)
hist5, test5 = train_and_eval_model(model5, train_loader, val_loader, test_loader, epochs=10, lr=0.0001)
plot_results(hist5, "Model 5: MobileNetV2 Learned")
```
